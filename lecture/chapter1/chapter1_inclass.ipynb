# 变量与基本语法

## 变量赋值与动态类型

* **动态类型**：Python 不需要显式声明类型，类型由“值”决定。
* **基本语法**：
```python
变量名 = 对象
```

## 实机演示

````python
sample_size = 500          # 整数（int）
print("sample_size 的值:", sample_size)
print("sample_size 的类型:", type(sample_size))

confidence_level = 0.95    # 浮点数（float）
print("confidence_level 的值:", confidence_level)
print("confidence_level 的类型:", type(confidence_level))

research_topic = "消费行为分析"  # 字符串（str）
print("research_topic 的值:", research_topic)
print("research_topic 的类型:", type(research_topic))
````

## 命名规范

在统计项目中，良好的命名习惯至关重要。对比以下两种写法，哪一种在三个月后还能让你一眼看懂代码的含义？

* **反面教材**：`a = 500`
* **推荐规范**：`sample_size = 500`

## PEP-8 命名规范

1.  **组成**：字母、数字、下划线，**不能数字开头**。
2.  **风格**：蛇形命名法(`snake_case`)。
3.  **大小写敏感**：`data` $\neq$ `Data`。
4.  **避免冲突**：
    * 严禁使用关键字 (`if`, `for`)。

| 规范性 | 示例 |
| :--- | :--- |
| $\checkmark$ | `gpa_score` |
| $\times$ | `1st_place` |
| $\times$ | `gpa score` |

## 新手的第一个“坑”：浮点数精度

````python
print("0.1 + 0.2 等于 0.3 吗？", 0.1 + 0.2 == 0.3)
print("实际计算结果:", 0.1 + 0.2)
````

## 正确的判断方式

````python
print("0.1 + 0.2 等于 0.3 吗？", abs(0.1 + 0.2 - 0.3) <= 1e-10)
````

## 字符串与格式化输出

### 格式化输出：**f-string**

````python
avg_score = 88.54321
print(f"该班级的平均分为：{avg_score:.2f}")
````

## 类型转换 (Casting)

  * `int(x)`：转换为整数
  * `float(x)`：转换为浮点数
  * `str(x)`：转换为字符串

### 链式转换演示

````python
age_str = "3.14"
age_int = int(float(age_str))
print(f"转换后整数: {age_int}")
````

## 经典 Debug
* 在处理统计调查数据时，我们经常会从 Excel 读取数据。
* 有时，看起来像数字的值，实际上是字符串（文本）。
* **文本不能直接与数字相加！**

````python
mean_score = "85.5"  
adjustment = 2.5
corrected_score = mean_score + adjustment
print(f"调整后的分数为: {corrected_score}")
````

# 运算符与逻辑表达式

## 算术运算符

除了四则运算 `+`、`-`、`*`、`/` 外，Python 还支持下面的算术运算符：

| 运算符 | 描述 | 示例 | 结果 |
| :-- | :-- | :-- | :-- |
| `**` | 幂运算 | `2 ** 3` | `8` |
| `%` | 取模 | `10 % 3` | `1` |
| `//` | 整除 | `10 // 3` | `3` |

## 比较运算与逻辑运算

* **比较运算符**：`==`, `!=`, `>`, `<`, `>=`, `<=`
* **逻辑运算符**：`and`（且）, `or`（或）, `not`（非）

例如，判断一位受访者的年龄是否处于 `18` 到 `60` 之间：

````python
age = 25
is_working_age = (age >= 18) and (age < 60)
print(f"是否处于指定年龄范围：{is_working_age}")
````
