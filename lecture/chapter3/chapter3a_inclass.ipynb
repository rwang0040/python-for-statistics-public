# 为什么需要 NumPy

## 列表在统计应用中的局限性

* 第二章已经建立了用列表存放一组成绩、一组年龄、一组观测值的基础。
* 列表很好用，但当一整批数据要执行同一个数值操作时，往往仍然需要借助 `for` 循环。
* 例如：全班 50 个人的成绩都想加 5 分，就需要逐个处理。

```python
# 这是列表
scores = [82, 76, 91, 88, 79]
# 得用循环才能整体加 5
print([score+5 for score in scores])
# 如果写成 scores + 5，会直接报错
```

## 列表到数组：一次升级

* NumPy 数组正是为这种“批量数值计算”而设计的。
* 它让“一整列数值”能够像一个整体那样被统一处理。

### 本节学习主线
NumPy 是 Python 中常用的第三方库，主要用于数值数组与数值计算。本节会介绍它的核心功能：把一列或一块数值数据放进 NumPy 数组中统一处理。很多统计计算换成数组表达后，会更自然。

# 从列表到数组

## 创建数组：给列表“升个级”

* NumPy 里最核心的对象是数组（`ndarray`）。
* 把已经熟悉的列表交给 `np.array(...)`，就可以得到数组。
* 打印出来虽然和列表很像，但它已经能够直接承担批量运算。

```python
import numpy as np

scores = [82, 76, 91, 88, 79]
score_array = np.array(scores)

print(score_array)
print(type(score_array))
```

## 数组看起来像什么

* 一维数组可以理解为“**一列数值**”。
* 这和统计学里的**一个变量**、**一组观测值**几乎是同一个意思。
* 也可以把数组理解为：把表格中的一列数值单独抽出来，放进一个更适合计算的结构中。

```python
import numpy as np

heights = np.array([168.5, 172.0, 165.3, 178.1])
print(heights)
```

## `shape`：看一眼数组的“身材”

* 拿到一个数组后，首先常常要知道它有多大、有几行几列。
* `shape` 就是用来描述数组结构的。
* 一维数组的 `shape` 写作 `(元素个数,)`。那个逗号表示这里只有一个维度。

```python
heights = np.array([168.5, 172.0, 165.3, 178.1])
print(f"shape: {heights.shape}")
```

## 二维数组的直观理解

* 如果一维数组是一列数据，那二维数组就像**多行多列的数据块**。
* 它还不是最完整的表格表达方式，但已经能帮助建立对表格数据的直观认识。

```python
import numpy as np

scores = np.array([
    [82, 76, 91],
    [88, 79, 85]
])

print(scores)
print(f"shape: {scores.shape}")
```

* 这里 `(2, 3)` 的意思是：2 行、3 列。

# 向量化计算

## 数组与标量运算：一次加给所有人

* 第二章里，如果想给每个成绩都加 5 分，通常需要写循环。
* 在数组里，这件事可以直接写成整体运算：**直接 `+ 5` 即可**。

```python
import numpy as np

scores = np.array([82, 76, 91, 88, 79])
adjusted_scores = scores + 5

print(f"原成绩: {scores}")
print(f"加分后: {adjusted_scores}")
```

## 数组与数组运算：对应位置直接算

* 如果两个数组长度一致，它们可以在**对应位置**做加减乘除。
* 这很适合表示“同一批样本上的两个指标”，例如基本工资和奖金。

```python
import numpy as np

income = np.array([4200, 5100, 3900])
bonus = np.array([300, 200, 500])

final_income = income + bonus
print(final_income)
```

## 向量化的直观含义

* 重点不在于记住“向量化”这个词。
* 更重要的是理解：**把一整列数据当成一个整体来操作，比逐个处理更贴近统计中的数值计算场景**。

## 对比：循环与向量化

* 把第二章的循环写法和数组写法放在一起看，差别会更直观。

```python
# 第二章的方式：循环
scores = [82, 76, 91, 88, 79]
adjusted_scores = []

for score in scores:
    adjusted_scores.append(score + 5)

print(f"循环结果: {adjusted_scores}")
```

```python
# NumPy 的方式：向量化
import numpy as np

scores = np.array([82, 76, 91, 88, 79])
print(f"向量化结果: {scores + 5}")
```

* 两种写法得到的结果一致，但数组写法更紧凑，而且更贴近“处理一列数据”的思路。

## 什么时候更适合用数组

* 数据主要是数值型（成绩、身高、收入、温度……）
* 同一个运算要作用到整批数据上
* 后面还要继续做统计计算（求均值、标准差、筛选）

### 不是所有数据都要转成数组
如果数据里混着文字、日期、类别，列表和字典仍然很有价值。数组最适合的是**纯数值**的批量处理。

# 数组的基础索引

## 一维数组取值

* 数组的取值规则和列表几乎一致。
* 第一项仍然是位置 `0`，最后一项仍然可以写成 `-1`。

```python
import numpy as np

scores = np.array([82, 76, 91, 88, 79])
print(f"第一个成绩: {scores[0]}")
print(f"最后一个成绩: {scores[-1]}")
```

## 切片

* 切片规则也和列表一致：`[起点:终点]`，左闭右开。

```python
import numpy as np

scores = np.array([82, 76, 91, 88, 79])
print(f"前三个成绩: {scores[:3]}")
print(f"后两个成绩: {scores[2:]}")
```

## 二维数组中的行列

* 二维数组的索引规则是：**先选行，再选列**，中间用逗号隔开。
* 需要先建立这个最基础的顺序：`[行, 列]`。

```python
import numpy as np

scores = np.array([
    [82, 76, 91],
    [88, 79, 85]
])

print(f"第一行: {scores[0]}")
print(f"第一行第二列: {scores[0, 1]}")
```

### 新手提醒
二维索引的基础写法是 `[行, 列]`。另外，**不要把行列顺序写反**，`[0, 1]` 和 `[1, 0]` 取到的是不同的值。

# 批量计算的意义

* 当数据是一批数值时，程序不一定总要逐个处理。
* 向量化不是炫技，而是**更贴近“处理一个变量”的统计直觉**。
* 把一列数据当成一个整体来操作，这种习惯会延续到 Pandas 和机器学习的数据预处理中。

## 计算效率、内存与核验

* 向量化写法通常更简洁，处理大量数值时也往往更高效。
* 数组仍然需要占用计算机内存；数据过大时，单机无法一次装下全部内容。
* 无论写法多快，都要检查 `shape`、数据类型、取值范围和部分结果。
* “代码成功运行”不等于“计算结果一定正确”。

# 本节小结

* NumPy 数组可以看作**列表的数值增强版**，适合表示一列数值型数据
* `shape` 可帮助快速了解数组的结构
* 向量化计算可实现**将一整列数值作为整体处理**
* 数组的索引和切片与第二章的列表规则保持连贯
