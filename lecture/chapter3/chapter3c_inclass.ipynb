# 使用随机数生成器生成样本

## 创建随机数生成器

* `np.random.default_rng(种子)` 可以创建一个随机数生成器。
* 下面统一记作 `rng`，后面所有随机样本都从这里生成。
* `rng.random()` 可以生成在 `[0, 1)` 上均匀分布的随机数。

```python
import numpy as np
rng = np.random.default_rng(202604)

random_value = rng.random()
print(f"单个随机数：{random_value}")
```

## 固定种子与可复现

* 使用相同种子，可以重新生成同一串随机数。
* 改变种子，会得到另一组随机结果。
* 固定种子不是取消随机性，而是让演示和分析可以复现。

```python
first = np.random.default_rng(202604).random(3)
second = np.random.default_rng(202604).random(3)

print(first)
print(second)
print("两次结果是否相同：", np.allclose(first, second))
```

## 一组随机数也是一组数据

* 除了单个随机数，还可以直接生成一组随机数。
* 这与前两节把数据放进数组统一处理的思路是一致的。

```python
random_values = rng.random(size=10)
print(f"一组随机数：{random_values}")
```

## 生成随机整数

* `rng.integers(low, high, size)` 可以生成位于 `[low, high)` 区间的一组随机整数。

```python
random_ints = rng.integers(0, 10, size=10)
print(f"随机整数：{random_ints}")
```

## 生成指定区间内的随机小数

* `rng.random(size)` 默认生成 `[0, 1)` 上的随机小数。
* 如果需要在其他区间中生成随机小数，可以使用 `rng.uniform(low, high, size)`。
* 例如，`rng.uniform(1, 2, size=10)` 会生成位于 `[1, 2)` 区间的一组随机小数。

```python
uniform_values = rng.uniform(1, 2, size=10)
print(f"[1, 2) 上的随机小数：{uniform_values}")
```

## 二项分布里的 `n` 和 `size`

* 单次抛一枚硬币，可以理解为一次伯努利试验。
* 如果关心“连续抛多次以后，正面一共出现了几次”，更常用二项分布来描述。
* `rng.binomial(n, p)` 中，`n` 表示一次实验里重复试验的次数，`p` 表示单次成功的概率。

```python
coin_result = rng.binomial(n=10, p=0.5)
print(f"抛 10 次硬币，出现正面的次数：{coin_result}")
```

* 如果把这样的实验再重复很多次，可以继续使用 `size` 参数。
* 这时 `size` 表示“把同一种实验重复多少次”。

```python
coin_results = rng.binomial(n=10, p=0.5, size=5)
print(f"将抛 10 次硬币的实验重复 5 次：{coin_results}")
```

## 正态分布示例

* 这里先建立“可以直接生成一组样本”的直觉。

```python
normal_sample = rng.normal(loc=70, scale=10, size=10)
print(f"正态样本：{normal_sample}")
```

# 抛硬币实验

## 一次实验：抛 20 次硬币

* 先看一次完整实验：连续抛 `20` 次硬币，记录出现正面的次数。

```python
coin_result = rng.binomial(n=20, p=0.5)
print(f"抛 20 次硬币，出现正面的次数：{coin_result}")
```

## 把次数换成比例

* 既然一次实验里抛了 `20` 次硬币，那么“出现正面的次数除以 20”就是这次实验的正面比例。

```python
coin_result = rng.binomial(n=20, p=0.5)
print(f"抛 20 次硬币，出现正面的比例：{coin_result / 20:.2f}")
```

## 把同一个实验重复多次

* 如果把“抛 20 次硬币”的实验重复很多次，就会得到一组实验结果。
* 数组中的每个元素表示一次完整实验里的正面个数。

```python
coin_results = rng.binomial(n=20, p=0.5, size=10)
print(f"10 次实验的正面个数：{coin_results}")
```

## 样本量变大时

* 当每次实验里抛硬币的次数增多时，平均正面比例通常会更稳定。

```python
for n in [10, 100, 1000]:
    coin_results = rng.binomial(n=n, p=0.5, size=10)
    print(f"将抛 {n} 次硬币的实验重复 10 次，正面比例的平均值：{(coin_results / n).mean():.2f}")
    print(f"将抛 {n} 次硬币的实验重复 10 次，正面比例的标准差：{(coin_results / n).std():.4f}")
```

# 样本量与估计稳定性

## 不只比例会更稳定

* 稳定性不只出现在“正面比例”上，也会出现在其他基于样本的数值估计上。
* 下面用样本均值做一个更一般的观察。

## 二维数组里的多次实验

* 注意这里把 `size` 写成了 `(10, 50)`。
* 这表示一次生成一个 `10 × 50` 的二维数组：可以把它理解为 **重复 10 次实验，每次实验生成 50 个随机数**。

```python
samples = rng.normal(loc=70, scale=10, size=(10, 50))
sample_means = samples.mean(axis=1)

print(f"10 次实验各自的样本均值：{sample_means}")
```

## 比较不同样本量

* 如果每次实验中的样本量增大，样本均值通常会更稳定。

```python
for n in [10, 100, 1000]:
    samples = rng.normal(loc=70, scale=10, size=(10, n))
    sample_means = samples.mean(axis=1)

    print(f"每次生成 {n} 个随机数并重复 10 次时，样本均值的均值是：{sample_means.mean():.2f}")
    print(f"每次生成 {n} 个随机数并重复 10 次时，样本均值的标准差是：{sample_means.std():.4f}")
```

# 综合案例：蒙特卡洛 (Monte Carlo) 的直观理解

## 什么是蒙特卡洛方法

* 如果一个问题可以转化成“随机点落在某个区域中的比例”，程序就有机会用随机模拟来近似求解。
* 这类方法常被称为 **蒙特卡洛方法**。
* 这里先建立最基础的技术直觉：用随机比例近似一个数值。

## 随机点与几何比例

* 在边长为 `1` 的正方形中，取一个四分之一圆，圆的半径也是 `1`。
* 正方形面积是 `1`，四分之一圆面积是 `pi / 4`。
* 如果在正方形中随机撒很多点，那么“落在四分之一圆内的点所占比例”就会接近 `pi / 4`。

## 先生成一批点

```python
x = rng.random(10)
y = rng.random(10)

print(f"x 坐标：{x}")
print(f"y 坐标：{y}")
```

## 判断点是否落在圆内

* 对于点 `(x, y)`，如果满足 `x**2 + y**2 <= 1`，就说明它落在四分之一圆内。

```python
x = rng.random(10)
y = rng.random(10)

inside = x**2 + y**2 <= 1
print(f"是否在圆内：{inside}")
```

## 用比例估计 `pi`

* `inside` 是一组真假值。
* 它的均值等于“落在圆内的点所占比例”。
* 因为这个比例接近 `pi / 4`，所以可以用 `4 * inside.mean()` 来估计 `pi`。

```python
x = rng.random(10)
y = rng.random(10)

inside = x**2 + y**2 <= 1
pi_est = 4 * inside.mean()

print(f"pi 的估计值：{pi_est}")
```
