# 布尔筛选

## 条件比较会得到什么

* 数组与单个条件比较时，返回的不是一个真假值，而是一组真假结果。
* 可以理解为：程序正在逐个位置判断是否满足条件。
* 这组真假结果常被称为 **布尔掩码**，英文里常写作 `mask`。

```python
import numpy as np

scores = np.array([82, 76, 91, 58, 79])
high_mask = scores >= 80

print(f"是否高于等于 80: {high_mask}")
```

## 用条件筛选数据

* 把这组真假结果放回原数组中，就能保留满足条件的元素。
* 这种写法在数据分析中极为常见。
* 先把条件结果保存为 `mask`，再用 `scores[mask]` 筛选，会让程序更容易读。

```python
scores = np.array([82, 76, 91, 58, 79])
high_mask = scores >= 80
high_scores = scores[high_mask]

print(f"掩码: {high_mask}")
print(f"筛选结果: {high_scores}")
```

## 例子：筛选成年样本

```python
ages = np.array([17, 18, 21, 16, 20, 19])
adult_mask = ages >= 18
adult_ages = ages[adult_mask]

print(f"成年掩码: {adult_mask}")
print(f"成年样本: {adult_ages}")
```

## 布尔筛选的意义

* 在 Pandas 和机器学习数据预处理中，“按条件保留部分数据”会反复出现。
* 重点在于掌握按条件筛选数据的思路。
* 对统计任务来说，掩码不只可以筛选数据，也可以用于统计人数、计算分组均值和处理缺失值。

## 组合多个掩码

* 统计任务中经常需要同时满足多个条件。
* 在 NumPy 数组中，组合多个条件时常用 `&` 表示“并且”，用 `|` 表示“或者”。
* 每个比较条件最好用括号包起来，例如 `(ages >= 18) & (income >= 5000)`。
* 不要用 Python 普通的 `and` / `or` 来组合数组条件。

```python
ages = np.array([17, 22, 35, 16, 48, 63])
income = np.array([4200, 5100, 6800, 3900, 5700, 4600])

adult_high_income_mask = (ages >= 18) & (income >= 5000)
adult_high_income = income[adult_high_income_mask]

print(f"成年且收入不低于 5000 的掩码: {adult_high_income_mask}")
print(f"对应收入: {adult_high_income}")
```

# 统计聚合

## 聚合的含义

* 聚合是把“一批数值”概括成“几个统计结果”。
* 常见的有总和、均值、最大值、最小值、标准差。

## 常用聚合函数

```python
scores = np.array([82, 76, 91, 58, 79])

print(f"总分: {scores.sum()}")
print(f"平均分: {scores.mean()}")
print(f"最高分: {scores.max()}")
print(f"最低分: {scores.min()}")
print(f"标准差: {scores.std()}")
```

## 缺失值与 `np.nan`

* 真实数据中经常会有缺失值。
* 在 NumPy 的数值数组中，常用 `np.nan` 表示缺失值。
* `np.isnan(array)` 可以判断哪些位置是缺失值；前面加 `~` 可以得到“非缺失”的掩码。
* 如果数组里有 `np.nan`，直接计算均值通常也会得到 `np.nan`。

```python
ages = np.array([25, 30, np.nan, 22, 18], dtype=float)
valid_age_mask = ~np.isnan(ages)

print(f"年龄数组: {ages}")
print(f"直接计算均值: {ages.mean()}")
print(f"非缺失掩码: {valid_age_mask}")
print(f"忽略缺失后的均值: {ages[valid_age_mask].mean():.2f}")
```

## 聚合与统计学习

* 这些统计量本身不复杂。
* 更重要的是建立一种习惯：数组不仅用于存放数据，也天然适合做统计汇总。

# 二维数组的按行与按列统计

## 二维数组可以表示什么

* 二维数组可以粗略理解为“多行样本、多个指标”。
* 虽然还不是最完整的表格表达方式，但已能承载很多基础统计计算。

```python
scores = np.array([
    [82, 76, 91],
    [88, 79, 85],
    [90, 81, 87]
])

print(f"原数组:\n{scores}")
```

## 整体均值

```python
scores = np.array([
    [82, 76, 91],
    [88, 79, 85],
    [90, 81, 87]
])

print(f"整体均值: {scores.mean():.2f}")
```

## 按行理解

* 如果把每一行看作一个学生的多门成绩，按行均值就是“每个学生的平均分”。

```python
scores = np.array([
    [82, 76, 91],
    [88, 79, 85],
    [90, 81, 87]
])

print(f"按行均值: {scores.mean(axis=1)}")
```

## 按列理解

* 如果把每一列看作一门课程，按列均值就是“每门课程的平均分”。

```python
scores = np.array([
    [82, 76, 91],
    [88, 79, 85],
    [90, 81, 87]
])

print(f"按列均值: {scores.mean(axis=0)}")
```

## 先建立方向感

### 当前只需掌握
将 `axis=0` 理解为“按列汇总”，`axis=1` 理解为“按行汇总”即可。本节先建立这个最基础的方向感。

# 广播的直观理解

## 数组与标量运算

* 上一节已经出现过数组加标量的写法。
* 可以先直观理解为：标量会自动应用到数组的每个元素上。

```python
scores = np.array([82, 76, 91, 88, 79])
print(f"统一加 5 分后: {scores + 5}")
```

## 二维数组与一维数组

* 当形状匹配时，一维数组可以自动作用到二维数组的每一行。
* 这里先观察一维数组如何自动作用到二维数组的每一行。

```python
scores = np.array([
    [82, 76, 91],
    [88, 79, 85],
    [90, 81, 87]
])

bonus = np.array([2, 3, 1])
print(f"加分后成绩:\n{scores + bonus}")
```

# 综合案例：筛选女性样本并计算平均收入

## 任务描述

* 已知一组收入数据
* 第一列表示性别：`0` 代表女性，`1` 代表男性
* 第二列表示实际收入
* 先筛选出女性对应的收入，再计算女性的平均收入

```python
import numpy as np

income = np.array([
    [0, 4200],
    [0, 5100],
    [1, 3900],
    [1, 6800],
    [0, 5700],
    [1, 4600]
])
female_income = income[income[:, 0] == 0, 1]

print(f"女性收入: {female_income}")
print(f"女性平均收入: {female_income.mean()}")
```

## 本例回顾

* 二维数组里可以同时放“类别编码”和“数值变量”
* 可以先按第一列条件筛选，再取第二列数据
* 聚合函数可直接对筛选后的数值列做统计汇总
* 这一流程已更接近后续表格数据分析中的常见任务
