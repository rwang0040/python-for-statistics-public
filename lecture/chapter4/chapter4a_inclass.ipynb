# 为什么需要 Pandas

## 从“多条记录”到“表格数据”

* 第二章中，我们用列表和字典保存一批受访者记录。
* 第三章中，我们把数值列放进 NumPy 数组里做批量计算。
* 但真实统计数据通常不是单独的一列，而是一张表：每一行是一条记录，每一列是一个变量。

### 从 NumPy 到 Pandas
第三章的 NumPy 更适合处理同类型数值数组，例如一组年龄、一组成绩或一组随机数。第四章的 Pandas 更适合处理真实统计数据表：一行一个个体，一列一个变量，变量可以同时包含数值、类别、文本和缺失值。

## 表格更接近统计数据的样子

* 问卷数据、成绩数据、实验数据、社会调查数据，通常都有“行”和“列”。
* 行表示观测对象或一条观测记录。
* 列表示变量，例如性别、年龄、城市、成绩、学习时长。
* Pandas 正是 Python 中最常用的表格数据处理工具。

### 本章学习主线
Pandas 是后续数据分析、可视化和统计建模的核心工具。接下来将围绕“读取—检查—清洗—汇总—合并”逐步完成常见的数据整理任务。

# `Series` 与 `DataFrame`

## `Series`：带标签的一列数据

* `Series` 可以理解成“一列带名字的数据”。
* 它比普通列表多了索引信息。

```python
import pandas as pd

ages = pd.Series([25, 30, 17, 22], name="age")
print(ages)
```

## `DataFrame`：多列放在一起

* `DataFrame` 可以理解成一张表。
* 每一列可以是一个变量，每一行可以是一条观测记录。

```python
records = [
    {"name": "Alice", "gender": "female", "age": 25},
    {"name": "Bob", "gender": "male", "age": 30},
    {"name": "Charlie", "gender": "male", "age": 17},
]

df = pd.DataFrame(records)
print(df)
```

## 和前面内容的关系

* 列表适合存放一组值。
* 字典适合描述一条记录。
* NumPy 数组适合批量数值计算。
* DataFrame 更适合处理多变量、多记录的结构化数据。

| 已有工具 | 更适合的场景 | 本章如何衔接 |
|---|---|---|
| 列表 | 一组简单数据 | 可以变成一列 `Series` |
| 字典 | 一条带字段名的记录 | 多条字典记录可以变成 `DataFrame` |
| NumPy 数组 | 同类型数值的批量运算 | Pandas 的数值列可以继续做向量化计算 |
| DataFrame | 多变量表格数据 | 本章的主要数据对象 |

# 读取 CSV 文件

## 为什么要从文件读取

* 真实数据通常不会直接写在代码里。
* 更常见的情况是：数据保存在 CSV、Excel 或数据库中。
* 本节先从最常见的 CSV 文件开始。

## `pd.read_csv()`

* `pd.read_csv(path)` 可以读取 CSV 文件，并返回一个 DataFrame。
* 读进来以后，就可以继续使用 Pandas 的方法观察和处理。
* `head()` 可以先查看前几行，确认数据是否读入成功。

```python
df = pd.read_csv(DATA_DIR / "survey_extended.csv")
print(df.head())
```

### 注意输出中的 `...`
在 HTML 幻灯片或 notebook 的输出区域中，如果 DataFrame 列数较多、显示空间不够，Pandas 可能会用 `...` 折叠中间几列。这只是显示被压缩，不表示这些列没有读入。要确认完整字段，可以查看 `df.columns`，或只选择几列单独输出。

### Excel 文件怎么办
在真实本地环境中，Pandas 也可以读取和导出 Excel 文件。例如 `pd.read_excel("data.xlsx")` 可以读取 Excel，`df.to_excel("result.xlsx", index=False)` 可以导出 Excel。读取 Excel 通常需要安装 `openpyxl` 等支持库。浏览器中的示例以更轻量、运行更稳定的 CSV 为主。

# 观察 DataFrame

## 前几行与整体形状

* 上一节已经用过 `head()`，它用于查看前几行。
* `shape` 用于查看行数和列数。

```python
df = pd.read_csv(DATA_DIR / "survey_extended.csv")

print(df.head())
print(df.shape)
```

## 字段名

* `columns` 可以查看所有列名。
* 读到陌生数据时，先看列名非常重要。

```python
print(df.columns)
```

## 数据类型与缺失概况

* `info()` 可以快速查看每列的数据类型和非缺失数量。
* 它常用于刚拿到数据时做第一轮检查。

```python
df.info()
```

## 数值列的描述统计

* `describe()` 会对数值列给出常见统计摘要。
* 包括样本量、均值、标准差、最小值、四分位数和最大值。

```python
print(df.describe())
```
