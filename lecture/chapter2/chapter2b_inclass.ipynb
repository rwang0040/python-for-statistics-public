# 选择结构：让程序会判断

## `if` 语句

* 选择结构的作用是根据条件执行不同的操作。
* 在统计场景中，它经常用于分组、分类和规则判断。

````python
score = 87

if score >= 60:
    print("成绩合格")
````

## `if-else`

````python
score = 52

if score >= 60:
    print("成绩合格")
else:
    print("成绩不合格")
````

## 多种情况：`if-elif-else`

````python
score = 87

if score >= 90:
    level = "优秀"
elif score >= 80:
    level = "良好"
elif score >= 60:
    level = "及格"
else:
    level = "不及格"

print("成绩等级:", level)
````

## 条件顺序会影响结果

* 多分支结构会从上到下检查条件。
* 先写高分段，再写低分段，通常更清楚。
* 逻辑错误不一定触发语法错误，却会直接改变分类结果。

````python
score = 95

if score >= 60:
    level = "及格"
elif score >= 90:
    level = "优秀"

print(level)
````

# 循环结构：让程序重复做事

## `for`循环

* `for`适合对一组数据逐个处理。
* 遍历列表是本章最常见的循环场景。

````python
scores = [82, 76, 91, 88, 79]

for score in scores:
    print(score)
````

## 累计求和

````python
scores = [82, 76, 91, 88, 79]
total_score = 0

for score in scores:
    total_score = total_score + score

mean_score = total_score / len(scores)
print(f"平均分为：{mean_score:.2f}")
````

## 循环与判断

* 很多任务需要在遍历过程中进行判断。
* “逐条检查—符合条件—执行操作”是常见的处理模式。

````python
scores = [82, 76, 91, 58, 79]

for score in scores:
    if score < 60:
        print(f"{score}分需要重点关注")
````

## 条件计数

````python
scores = [82, 76, 91, 58, 79]
passed_count = 0

for score in scores:
    if score >= 60:
        passed_count = passed_count + 1

print("及格人数:", passed_count)
````

## `while`循环

* `while`适合重复次数不确定、但结束条件明确的任务。
* 入门阶段以认识其基本结构和防止无限循环为主。

````python
count = 1

while count <= 5:
    print("当前计数:", count)
    count = count + 1
````

## `for`和`while`怎么选

| 场景 | 更适合 |
| :-- | :-- |
| 已知要遍历一组数据 | `for` |
| 不确定重复次数，只知道结束条件 | `while` |

# 列表推导式：简洁地生成列表

## 基本形式

* 列表推导式可以把“循环生成新列表”写得更紧凑。
* 先理解普通循环，再学习等价的简洁写法。

```python
[表达式 for 变量 in 序列]
```

## 转换数据

````python
scores = [82, 76, 91, 88, 79]
adjusted_scores = [score + 5 for score in scores]
print(adjusted_scores)
````

## 筛选数据

````python
scores = [82, 76, 91, 58, 79]
passed_scores = [score for score in scores if score >= 60]
print("及格成绩:", passed_scores)
````

### 先保证可读
本节只使用基础列表推导式。遇到难以一眼读懂的写法时，优先改用普通循环。

# 综合案例：筛选多条受访者记录

````python
respondents = [
    {"name": "张三", "age": 16},
    {"name": "李四", "age": 21},
    {"name": "王五", "age": 19},
    {"name": "赵六", "age": 15}
]

adult_names = []

for respondent in respondents:
    if respondent["age"] >= 18:
        adult_names.append(respondent["name"])

print("成年人姓名:", adult_names)
print("成年人数:", len(adult_names))
````

## 改写成列表推导式

````python
adult_names = [
    respondent["name"]
    for respondent in respondents
    if respondent["age"] >= 18
]

print(adult_names)
````
