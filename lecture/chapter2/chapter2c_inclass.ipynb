# 自定义函数

## 最基本的函数写法

* 函数可以理解为“打包好的一段任务”。
* 调用时给它输入，它就按规则完成计算并返回结果。

```python
def 函数名(参数):
    函数体
    return 返回值
```

## 例子：计算两个数的平均值

* 这里先从最小例子开始，观察函数的结构。
* 先有参数，再有计算过程，最后返回结果。

````python
def mean_of_two_numbers(a, b):
    result = (a + b) / 2
    return result

print(mean_of_two_numbers(80, 90))
````

## 参数与返回值

* 参数：把外部数据传给函数。
* 返回值：把函数计算结果交回去。

````python
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi

student_bmi = calculate_bmi(68.5, 1.75)
print(f"BMI 为：{student_bmi:.2f}")
````

## 没有 `return` 会怎样

* 有些函数只负责输出，有些函数要把结果交回来。
* 如果没有 `return`，默认返回的是 `None`。
* 这也是很多初学者第一次分清 `print()` 和 `return` 的地方。

````python
def say_hello(name):
    print(f"你好，{name}")

result = say_hello("统计学同学")
print("函数返回值是：", result)
````

## 函数能带来什么

* 代码更短
* 逻辑更清楚
* 后续调试更方便

# 用函数处理一组数据

## 例子：计算受访者年龄列表平均值

* 前面我们处理的是单个数。
* 现在开始让函数处理“一组数据”。

````python
def calculate_mean(ages):
    total_age = sum(ages)
    mean_age = total_age / len(ages)
    return mean_age

ages = [18, 21, 19, 22, 20]
print(f"平均年龄为：{calculate_mean(ages):.2f}")
````

## 函数名要有意义

* `f1()` 不利于阅读
* `calculate_mean()` 一看就知道功能

## 函数不是魔法盒子

### 不要把函数当成“魔法盒子”
写函数时仍然要清楚输入是什么、输出是什么、内部做了什么。

## 把任务拆成几个函数

* 当一个程序里有多个任务时，可以拆成多个函数分别完成。
* 这样每个函数都更短，也更容易检查。
* 对数据分析来说，这一步很重要，因为“读取、计算、判断”本来就是不同任务。

````python
def calculate_mean(ages):
    return sum(ages) / len(ages)

def count_adults(ages):
    adult_count = 0
    for age in ages:
        if age >= 18:
            adult_count = adult_count + 1
    return adult_count

ages = [16, 18, 21, 17, 24]
print(f"平均年龄：{calculate_mean(ages):.2f}")
print(f"成年人数量：{count_adults(ages)}")
````

## 参数传递的直观理解

* 参数的作用，就是把外部数据交给函数处理。
* 调用时传什么值，函数就拿到什么值。

````python
def add_years(age, years):
    return age + years

next_year_age = add_years(20, 1)
print("明年的年龄:", next_year_age)
````

# 异常处理：程序为什么会报错

## 报错并不可怕

* 入门阶段最重要的不是“永不报错”。
* 而是学会识别错误、定位错误、处理错误。

## 转换时报错

* 报错并不意味着程序“坏掉了”。
* 它只是说明：当前输入不符合这段代码的要求。
* 在真实数据里，这种情况非常常见——比如年龄一栏里填的不是 `25`，而是 `二十五` 或 `25岁`。

````python
value = "3.14"
print(int(value))
````

## `try-except` 基本结构

```python
try:
    可能出错的代码
except ValueError:
    数值转换失败后的处理方式
```

## 用 `try-except` 处理转换错误

* `try` 里放可能出错的代码。
* `except` 里放出错后的处理方式。

````python
value = "3.14"

try:
    number = int(value)
    print(number)
except ValueError:
    print("转换失败：这个字符串不能直接转成整数。")
````

## 更合理的写法

* 有时候不是数据完全错了，而是转换方式选错了。
* 这时就要重新判断，问题到底出在输入还是写法。
* 记住：报错不是终点，而是帮我们定位问题的线索。

````python
value = "3.14"

try:
    number = float(value)
    print("转换成功：", number)
except ValueError:
    print("输入数据格式不正确。")
````

## 异常处理不是忽略问题

### 不要滥用 `except`
异常处理的目标是帮助定位问题，而不是把问题直接“糊过去”。能够判断错误类型时，应捕获具体异常；例如数值转换失败通常捕获 `ValueError`。

# 综合案例：把几个函数接起来

* 一个完整程序往往不是只靠一个函数完成。
* 更常见的写法是：不同函数分别负责不同任务。

````python
def calculate_mean_age(ages):
    return sum(ages) / len(ages)

def has_minor(ages):
    for age in ages:
        if age < 18:
            return True
    return False

def main():
    ages = [16, 18, 21, 17, 24]
    mean_age = calculate_mean_age(ages)
    minor_flag = has_minor(ages)

    print(f"平均年龄为：{mean_age:.2f}")
    print(f"是否有未成年人：{minor_flag}")

main()
````

## 程序结构

* 一个函数负责一种任务
* `main()` 负责把流程串起来
* 这样写更清楚，也更容易检查和修改
* 这种“分工明确”的写法，会在下节课的文本清洗任务里继续用到。
