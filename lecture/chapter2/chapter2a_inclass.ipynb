# 列表：存放一组数据

## 列表的基本概念

* 列表（`list`）适合表示一组有顺序的数据。
* 在统计场景中，它常常可以表示一个样本序列、一组观测值或若干类别标签。
* 你也可以把它理解成“某个指标的一列观测值”，例如一组成绩、一组年龄或一组身高。

```python
scores = [82, 76, 91, 88, 79]
print(scores)
print(type(scores))
```

## 索引与取值

* Python 使用从 `0` 开始的索引。
* 第一个元素是 `scores[0]`，最后一个元素也可以写成 `scores[-1]`。

```python
scores = [82, 76, 91, 88, 79]
print("第一个成绩:", scores[0])
print("最后一个成绩:", scores[-1])
```

## 列表切片

* 切片可以帮助我们一次取出一部分数据。
* 在数据处理中，这种操作非常常见。

```python
scores = [82, 76, 91, 88, 79]
print("前3个成绩:", scores[:3])
print("后2个成绩:", scores[3:])
```

## 常见操作

* 列表不仅能存数据，还能直接配合一些基础函数做简单统计。
* 例如 `len()`、`max()`、`min()`、`sum()`。

```python
scores = [82, 76, 91]
scores.append(88)
scores.append(79)
print(scores)
print("样本量:", len(scores))
print("最高分:", max(scores))
print("最低分:", min(scores))
print("总分:", sum(scores))
```

## 索引越界

### 新手提醒
列表索引从 `0` 开始，不是从 `1` 开始。
如果列表长度是 5，那么最后一个合法位置是 `4`，不是 `5`。

```python
scores = [82, 76, 91, 88, 79]
print(scores[5])
```

## 统计语境中的列表

* 一个列表可以看成一列原始观测值。
* 但列表本身不会自动进行统计分析，我们仍然需要写逻辑来处理它。

```python
scores = [82, 76, 91, 88, 79]
mean_score = sum(scores) / len(scores)
print(f"平均分为：{mean_score:.2f}")
```

# 字典：用名字对应信息

## 字典的基本概念

* 字典（`dict`）适合表示“键-值对应关系”。
* 在统计与调查数据中，它适合保存结构化信息。
* 如果说列表更像“一列数据”，那么字典更像“一条观测记录的完整画像”。

```python
student = {
    "name": "张三",
    "gender": "女",
    "score": 89,
    "major": "统计学"
}

print(student)
print(type(student))
```

## 按键取值

* 字典中的每个键都对应一类信息。
* 通过键名取值，代码会比按位置取值更清楚。

```python
student = {
    "name": "张三",
    "gender": "女",
    "score": 89,
    "major": "统计学"
}

print("姓名:", student["name"])
print("成绩:", student["score"])
```

## 修改字典内容

* 字典里的值不是固定不变的。
* 在数据处理过程中，我们经常需要补充或更新字段。

```python
student = {"name": "张三", "score": 89}
student["score"] = 92
student["major"] = "统计学"

print(student)
```

## 列表和字典

* 列表更像“一组值”。
* 字典更像“一条记录”。
* 后面学习 `DataFrame` 时，可以把它理解成“许多记录排在一起”。

## 一条记录

```python
survey = {
    "age": 19,
    "city": "上海",
    "height": 168.5
}

print(f"年龄：{survey['age']}")
print(f"城市：{survey['city']}")
```

## 嵌套结构：更接近真实数据

* 真实数据往往不是“只有一层”的。
* 例如：多个学生记录，可以表示为“列表里放多个字典”。

```python
students = [
    {"name": "张三", "score": 85},
    {"name": "李四", "score": 92},
    {"name": "王五", "score": 78}
]

print(students[0]["name"])
print(students[1]["score"])
```
