## Jupyter使用提示

1. Notebook页面顺序不等于实际执行顺序，变量状态由代码单元的运行历史决定。
2. 修改代码后需要重新运行相应单元，屏幕上的旧输出不会自动更新。
3. 遇到变量找不到或结果对不上时，先检查是否漏运行或乱序运行。
4. 可以使用`Shift + Enter`运行当前单元，使用`Tab`尝试代码补全。
5. 如果内核状态混乱，重启内核后从头运行通常比反复猜测更可靠。
